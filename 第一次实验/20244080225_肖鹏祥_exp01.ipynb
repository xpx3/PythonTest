### 实验 1：基本运算、数据类型与字符串

本部分比较常用运算符，检查四种基础数据类型，并练习类型转换和字符串操作。

```python
print("8 / 3  =", 8 / 3, type(8 / 3))
print("8 // 3 =", 8 // 3, type(8 // 3))
print("8 % 3  =", 8 % 3, type(8 % 3))
print("2 ** 7 =", 2 ** 7, type(2 ** 7))
```

`/` 是普通除法，结果通常为浮点数；`//` 是向下取整的整除；`%` 求余数；`**` 表示乘方。

```python
values = [
    2024,          # 整数
    3.14,          # 浮点数
    "Python",      # 字符串
    True,          # 布尔值
]

for value in values:
    print(f"{value!r:>10} -> {type(value)}")
```

类型转换时，`int("123")` 把数字字符串转为整数，`float("3.14")` 转为浮点数，
`int(3.9)` 会直接截断小数部分。`int("goat")` 无法转换，因为字符串内容不是整数。

```python
print('int("123")   =', int("123"), type(int("123")))
print('float("3.14") =', float("3.14"), type(float("3.14")))
print("int(3.9)      =", int(3.9), type(int(3.9)))

try:
    int("goat")
except ValueError as error:
    print(f'int("goat") 引发异常：{error}')
```

字符串处理会生成新字符串；`strip()` 去除首尾空白，`lower()` 和 `upper()` 转换大小写，
切片提取部分字符，`split()` 拆分单词，`join()` 再把单词连接起来。

```python
text = " Python Data Analysis "
stripped = text.strip()

print("原字符串：", repr(text))
print("去除首尾空白：", repr(stripped))
print("小写：", stripped.lower())
print("大写：", stripped.upper())
print("切片 [2:6]：", stripped[2:6])

words = stripped.split()
print("拆分单词：", words)
print("用 - 连接：", "-".join(words))
```

字符串可以用 `+` 拼接、用 `*` 重复。字符串不可原地修改单个字符；
若要修改，需要构造一个新的字符串。

```python
text = "Python"
print("拼接：", text + " Lab")
print("重复：", "Hi! " * 3)

try:
    text[0] = "J"
except TypeError as error:
    print("直接修改字符失败：", error)

new_text = "J" + text[1:]
print("构造新字符串：", new_text)
```

### 实验 2：函数、条件、循环与递归

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in [95, 85, 60, 59, 0, 100]:
    print(f"{score} 分：{score_level(score)}")

try:
    score_level(101)
except ValueError as error:
    print("分数越界：", error)
```

```python
total_for = 0
for number in range(1, 101):
    total_for += number

total_while = 0
number = 1
while number <= 100:
    total_while += number
    number += 1

print("for 循环求和：", total_for)
print("while 循环求和：", total_while)
print("结果相同：", total_for == total_while == 5050)
```

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for factor in range(2, n + 1):
        result *= factor
    return result


for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)
    print(
        f"{n}!：递归={recursive_result}, 迭代={iterative_result}, "
        f"相同={recursive_result == iterative_result}"
    )
```

```python
def summary(values):
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return minimum, maximum, average


minimum, maximum, average = summary([10, 20, 30])
print("最小值、最大值、平均值：", minimum, maximum, average)
```

```python
def sum_all(*args):
    return sum(args)


print("无参数：", sum_all())
print("单个参数：", sum_all(10))
print("多个参数：", sum_all(1, 2, 3, 4, 5))
```

### 实验 3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_dict = dict(zip(names, scores))
print("姓名到成绩的字典：")
for name, score in score_dict.items():
    print(name, score)

print("带序号输出：")
for index, (name, score) in enumerate(score_dict.items(), start=1):
    print(index, name, score)

passed_names = [name for name, score in score_dict.items() if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("及格学生：", passed_names)
print("成绩平方：", squared_scores)
```

```python
count = len(scores)
average_score = sum(scores) / count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = sum(score >= 60 for score in scores) / count

print(f"人数：{count}")
print(f"平均分：{average_score:.2f}")
print(f"最高分：{highest_score}")
print(f"最低分：{lowest_score}")
print(f"及格率：{pass_rate:.2%}")
```

### 实验 4：文件读写、异常处理与模块

```python
import os

print("当前工作目录：", os.getcwd())
data_dir = os.path.join(".", "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")

student_text = '''name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete'''

with open(students_path, "w", encoding="utf-8") as file:
    file.write(student_text)

with open(students_path, "r", encoding="utf-8") as file:
    header = next(file).strip()
    print("表头：", header)
    for line in file:
        print("解析前：", line.strip().split(","))
```

`parse_student()` 负责把一行文本转成字典；字段数、姓名、整数成绩和成绩范围任何一项不合格，
都抛出 `ValueError`，便于调用方逐行处理错误。

```python
def parse_student(line):
    """将一行 name,score 解析为字典；无效记录抛出 ValueError。"""
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")

    name = parts[0].strip()
    score_text = parts[1].strip()

    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩必须为整数，当前为 {score_text!r}") from None

    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在 0 到 100 之间，当前为 {score}")

    return {"name": name, "score": score}
```

```python
valid_records = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as file:
    next(file)  # 跳过表头
    for line_number, line in enumerate(file, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append((line_number, str(error)))
            print(f"第 {line_number} 行无效：{error}")

print("有效记录：", valid_records)
print("有效记录数：", len(valid_records))
print("无效记录数：", len(invalid_records))

assert len(valid_records) == 6
assert len(invalid_records) == 3
```

```python
valid_count = len(valid_records)
invalid_count = len(invalid_records)
valid_average = sum(item["score"] for item in valid_records) / valid_count
valid_pass_rate = (
    sum(item["score"] >= 60 for item in valid_records) / valid_count
)

report_path = os.path.join(data_dir, "report.txt")

# w 模式先清空文件，再写入新内容。
with open(report_path, "w", encoding="utf-8") as file:
    file.write("旧内容：这一行会被下一次 w 写入覆盖。\n")

report_text = (
    f"有效记录数：{valid_count}\n"
    f"无效记录数：{invalid_count}\n"
    f"平均分：{valid_average:.2f}\n"
    f"及格率：{valid_pass_rate:.2%}\n"
)
with open(report_path, "w", encoding="utf-8") as file:
    file.write(report_text)

with open(report_path, "r", encoding="utf-8") as file:
    print("w 模式覆盖后的报告：")
    print(file.read())

# a 模式保留原内容，在文件末尾追加。
with open(report_path, "a", encoding="utf-8") as file:
    file.write("追加演示：a 模式保留了原报告。\n")

with open(report_path, "r", encoding="utf-8") as file:
    print("a 模式追加后的报告：")
    print(file.read())
```

### 实验 5：类、继承与对象复制

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score})"


student_objects = [
    Student(record["name"], record["score"]) for record in valid_records
]
for student in student_objects:
    print(student, "及格：", student.is_passed())
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"GraduateStudent(name={self.name!r}, score={self.score}, "
            f"research_topic={self.research_topic!r})"
        )


graduate = GraduateStudent("Alice", 85, "机器学习")
print(graduate)
print("是否及格：", graduate.is_passed())
```

```python
import copy


def copy_experiment(label, copy_function):
    original = {"name": "Alice", "scores": [85, 90]}
    copied = copy_function(original)
    copied["scores"].append(95)
    print(
        f"{label}：original={original['scores']}, "
        f"copied={copied['scores']}"
    )
    return original, copied


direct_original, direct_copy = copy_experiment("直接赋值", lambda item: item)
shallow_original, shallow_copy = copy_experiment("浅复制", copy.copy)
deep_original, deep_copy = copy_experiment("深复制", copy.deepcopy)

assert direct_original is direct_copy
assert shallow_original is not shallow_copy
assert shallow_original["scores"] is shallow_copy["scores"]
assert deep_original is not deep_copy
assert deep_original["scores"] is not deep_copy["scores"]
assert deep_original["scores"] == [85, 90]
```
