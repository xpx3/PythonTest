### 欧几里得算法

利用 `gcd(m, n) = gcd(n, m % n)`，当余数为 0 时，最后一个非零除数就是最大公约数。
这里先把输入转成非负整数，因此负数输入也会返回符合数学习惯的非负 GCD。

```python
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m


pairs = [(2018, 2019), (1200, 300), (5040, 60)]
for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")

print("负数测试：")
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-2018, -2019) =", gcd(-2018, -2019))
```

**负数行为说明：** 函数通过 `abs()` 统一处理输入，所以传入一个或两个负数时仍返回非负的 GCD。
这符合“最大公约数通常取非负整数”的数学习惯。

### 近似欧拉数 e

公式 (1) 为 `e = 1/0! + 1/1! + 1/2! + ...`；
函数从 `n = 1` 开始计“项数”，因此 `euler_infinite_sum(4)` 包含第 0 到第 3 项。

```python
import math


def euler_limit(n):
    if n <= 0:
        raise ValueError("n 必须是正整数")
    return float((1 + 1 / n) ** n)


def euler_infinite_sum(n):
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
    return float(total)


def euler_approx(epsilon):
    if epsilon <= 0:
        raise ValueError("epsilon 必须大于 0")

    target = math.exp(1)
    total = 0.0
    term = 1.0
    k = 0

    while abs(target - total) >= epsilon:
        total += term
        k += 1
        term /= k

    return float(total)


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"{k}: {euler_infinite_sum(k):.12f}")


print("euler_limit(1_000_000) =", euler_limit(1_000_000))
print("euler_infinite_sum(0)  =", euler_infinite_sum(0))
print("euler_infinite_sum(1)  =", euler_infinite_sum(1))
print("euler_infinite_sum(4)  =", euler_infinite_sum(4))
print("euler_approx(1e-10)    =", euler_approx(1e-10))
print("实际 e                 =", math.exp(1))
print("前 5 项近似表：")
print_euler_sum_table(5)

assert euler_infinite_sum(0) == 0.0
assert euler_infinite_sum(1) == 1.0
assert abs(euler_infinite_sum(4) - 2.6666666666666665) < 1e-12
assert abs(euler_limit(1_000_000) - math.exp(1)) < 1e-5
assert abs(euler_approx(1e-10) - math.exp(1)) < 1e-10
```

### 测试整数属性

`is_power_of_2` 反复除以 2；若最终得到 1，则该数是 2 的幂。
`is_power(b, n)` 使用相同的思路，反复除以底数 `b`。

```python
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1


print("2 的幂测试：")
for value in [1, 2, 8, 10, 16, 18]:
    print(f"is_power_of_2({value}) = {is_power_of_2(value)}")

print("一般幂测试：")
tests = [(3, 1), (3, 9), (3, 10), (5, 125), (5, 126)]
for base, value in tests:
    print(f"is_power({base}, {value}) = {is_power(base, value)}")

assert is_power_of_2(1) is True
assert is_power_of_2(1024) is True
assert is_power_of_2(1000) is False
assert is_power(3, 81) is True
assert is_power(3, 80) is False
assert is_power(1, 1) is True
assert is_power(1, 2) is False
```
